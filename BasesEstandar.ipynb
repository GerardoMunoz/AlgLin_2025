# **1. Axiomas de Espacio Vectorial**

Un espacio vectorial \\(V\\) sobre \\(\mathbb{R}\\) es un conjunto cuyos elementos pueden sumarse y multiplicarse por escalares, cumpliendo **10 axiomas**.

Sean \\(u,v,w\in V\\) y \\(a,b\in\mathbb{R}\\).

| #  | Axioma                               | Propiedad matemática  |
| -- | ------------------------------------ | --------------------- |
| 1  | Cerradura bajo la suma               | \\(u+v\in V\\)        |
| 2  | Conmutatividad                       | \\(u+v=v+u\\)         |
| 3  | Asociatividad                        | \\((u+v)+w=u+(v+w)\\) |
| 4  | Vector cero                          | \\(u+0=u\\)           |
| 5  | Inverso aditivo                      | \\(u+(-u)=0\\)        |
| 6  | Cerradura bajo producto escalar      | \\(au\in V\\)         |
| 7  | Identidad escalar                    | \\(1u=u\\)            |
| 8  | Asociatividad escalar                | \\(a(bu)=(ab)u\\)     |
| 9  | Distributividad respecto a vectores  | \\(a(u+v)=au+av\\)    |
| 10 | Distributividad respecto a escalares | \\((a+b)u=au+bu\\)    |

**Observación:** Los primeros cinco axiomas describen la estructura de la suma. Los últimos cinco relacionan la multiplicación por escalares con la suma.

# **2. Espacios vectoriales más comunes y sus bases estándar**

## **A. Espacios de coordenadas:&#x20;**\\(\mathbb{R}^n\\)

Espacio fundamental

Sus elementos son vectores de \\(n\\) componentes reales.

Por ejemplo:

$$ \mathbb{R}^2=\left\{ \begin{bmatrix}x \\ y \end{bmatrix}:x,y\in\mathbb R \right\} $$

Su base estándar es:

$$ \mathcal B= \left\{ \begin{bmatrix}1\\0\end{bmatrix}, \begin{bmatrix}0\\1\end{bmatrix} \right\} $$

Para \\(\mathbb R^3\\):

$$ \mathcal B= \left\{ \begin{bmatrix}1\\0\\0\end{bmatrix}, \begin{bmatrix}0\\1\\0\end{bmatrix}, \begin{bmatrix}0\\0\\1\end{bmatrix} \right\} $$

En general, la base estándar de \\(\mathbb R^n\\) es:

\\[ \mathcal B=\{e_1,e_2,\ldots,e_n\} \\]

Donde cada vector tiene un 1 en una posición y ceros en las demás.

## **B. Espacios de matrices:&#x20;**\\(M_{m\times n}(\mathbb R)\\)

Sus elementos son matrices de \\(m\\) filas y \\(n\\) columnas.

Por ejemplo, \\(M_{2\times2}(\mathbb R)\\):

\\[ A=\begin{bmatrix}a&b\\c&d\end{bmatrix} \\]

Su base estándar es:

\\[ \mathcal B= \left\{ \begin{bmatrix}1&0\\0&0\end{bmatrix}, \begin{bmatrix}0&1\\0&0\end{bmatrix}, \begin{bmatrix}0&0\\1&0\end{bmatrix}, \begin{bmatrix}0&0\\0&1\end{bmatrix} \right\} \\]

## **C. Espacios de polinomios de grado $n$ o menor:**\\(P_n\\)

Son los polinomios de grado menor o igual a \\(n\\).

Por ejemplo:

\\[ P_2=\{a_0+a_1x+a_2x^2:a_i\in\mathbb R\} \\]

Su base estándar es:

\\[ \mathcal B=\{1,x,x^2\} \\]



En general:

\\[ P_n=\operatorname{span}\{1,x,x^2,\ldots,x^n\} \\]

Y su dimensión es:

\\[ \dim(P_n)=n+1 \\]

## **D. Espacios de polinomios:**\\(P\\)

Es el conjunto de todos los polinomios.

Por ejemplo:

\\[ P_2=\{a_0+a_1x+a_2x^2:a_i\in\mathbb R\} \\]

Su base estándar tiene infinitos elementos:

\\[ \mathcal B=\{1,x,x^2,\ldots\} \\]

\\[ P_n=\operatorname{span}\{1,x,x^2,\ldots\} \\]

## **E. Espacios de funciones**

Un ejemplo común es el espacio \\(C[a,b]\\), formado por todas las funciones continuas en el intervalo \\([a,b]\\).

Ejemplos de sus elementos:

\\[ f(x)=x^2,\quad g(x)=\sin(x),\quad h(x)=e^x \\]

Este espacio tiene dimensión infinita.

Importante: no tiene una base estándar finita como \\(\mathbb R^n\\). Su base de Hamel requiere infinitos elementos y no se identifica simplemente con una lista de funciones habituales.

# **4. La conexión más importante: base y combinación lineal**

Una base de un espacio vectorial es un conjunto de vectores que cumple dos condiciones:

1. Es linealmente independiente.
2. Genera todo el espacio vectorial.

Esto significa que cualquier vector del espacio se puede expresar como una combinación lineal única de los elementos de la base.

Por ejemplo, en \\(\mathbb R^2\\):

\\[ v=\begin{bmatrix}3\\5\end{bmatrix} \\]

Se puede expresar como:

\\[ v=3\begin{bmatrix}1\\0\end{bmatrix} +5\begin{bmatrix}0\\1\end{bmatrix} \\]

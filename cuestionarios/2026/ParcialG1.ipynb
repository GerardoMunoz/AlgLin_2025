# Punto 1
> (40% del parcial)

Resolver el siguiete sistema de ecuaciones dado por la siguiente matriz extendida $$[A:\vec{b}]$$

```python
A.row_join(b)
```

1.a Lo primero es encontrar el determinante de la matriz $A=$

```python
A
```

utilizando cofactores por algún renglon. Se selecciona el último renglón debido a que es el renglón con más ceros.

$$|A|=a_{31}C_{31}+a_{32}C_{32}+a_{33}C_{33}$$
$$=a_{31}M_{31}-a_{32}M_{32}+a_{33}M_{33}$$
$$=0-(-1)\begin{vmatrix}1& 3\\4 &6 \end{vmatrix}+(1)\begin{vmatrix}1&-2 \\ 4& 5\end{vmatrix}$$
$$=0-(-1)(6-12)+(1)(5+8)$$
$$=-6+13=7$$

```python
A.det()
```

1.b Ahora se encuentra el determinante de la matriz $Ax=$

```python
Ax=b.row_join(A[:,1:])
Ax
```

utilizando cofactores por alguna columna. Se selecciona la segunda columna ya que ninguna tiene ceros (cualquier columna es viable).

$$|A|=a_{12}C_{12}+a_{22}C_{22}+a_{32}C_{32}$$
$$=a_{12}M_{12}-a_{22}M_{22}+a_{32}M_{32}$$
$$=-(-2)\begin{vmatrix}32& 6\\1 &1 \end{vmatrix}+5\begin{vmatrix}6& 3\\1 &1 \end{vmatrix}-(-1)\begin{vmatrix}6&3 \\ 32& 6\end{vmatrix}$$
$$=-(-2)(32-6)+5(6-3)-(-1)(36-96)$$
$$=52+15-60=7$$

y por lo tanto

$$x=\frac{A_x}{A}=\frac{7}{7}=1$$

```python
Ax.det(), Ax.det()/A.det()
```

1.c Ahora se encuentra el determinante de la matriz  Ay=

```python
Ay=A[:,0].row_join(b).row_join(A[:,2])
Ay
```

utilizando Sarrus.
$$\left[\begin{matrix}1 & 6 & 3\\4 & 32 & 6\\0 & 1 & 1\end{matrix}\right]\left.\begin{matrix}1 & 6 \\4 & 32 \\0 & 1 \end{matrix}\right.$$

$$|A_y|=32+0+12-0-6-24=14$$
entonces

$$y=\frac{A_y}{A}=\frac{14}{7}=2$$

```python
Ay.det(), Ay.det()/A.det()
```

1.d Ahora se encuentra el determinante de la matriz $A_z=$

```python
Az=A[:,:2].row_join(b)
Az
```

utilizando la eliminación de renglones de Gauss


```python
M=Az[:,:] #No afecta el determinante (1)
M
```

```python
M[1,:]-=4*M[0,:] #No afecta el determinante (1)
M
```

```python
M.row_swap(2,1) # Mulptiplica por (-1)
```

```python
M[2,:]+=13*M[1,:] #No afecta el determinante (1)(-1)
M
```

Determinante es el cambio de signo debido al intercambio de renglones por la multiplicacion de la diagonal

$$|A_z|=(-1)((1)(-1)(21))=21$$
$$z=\frac{A_z}{A}=\frac{21}{7}=3$$

```python
Az.det(), Az.det()/A.det()
```

# Taller: ortogonalidad, ortonormalidad y resolución de $Ax=b$

## Ejemplo 1, la transpuesta es la solución:

```python
A1 = sp.Matrix([[4/5, -3/5],[3/5,4/5]])
A1
```

```python
b1=sp.Matrix([1.4,-0.2])
b1
```


Vamos a encontrar el valor de $\vec{x}$ Que resuelve el siguiente sistema:
$$\left[\begin{matrix}0.8 & -0.6\\0.6 & 0.8\end{matrix}\right]\vec{x}=\left[\begin{matrix}1.4\\-0.2\end{matrix}\right]$$
Será que $$\vec{x}=A^T\vec{b}$$

```python
x=A1.T@b1
x
```

Sorprendentemente este la transpuesta resuelve el sistema

```python
A1@x
```

## Ejemplo 2, la proyección es la solución:

```python
A2 = sp.Matrix([[8, -6],[6,8]])
A2
```

```python
b2=sp.Matrix([14,-2])
b2
```

En este caso la transpuesta no es la solución

```python
x_mala=A2.T@b2
x_mala
```

Pero si podemos encontrar la respuesta de cada variable utilizando su respectiva columna para calcular sobre ella la proyección.

```python
x0= b2.dot(A2[:,0])/A2[:,0].dot(A2[:,0])
x0
```

```python

x1= b2.dot(A2[:,1])/A2[:,1].dot(A2[:,1])
x1
```

```python
A1@sp.Matrix([x0,x1])
```

## Ejemplo 3, solo sirve resolver el sistema de ecuaciones:

```python
A3 = sp.Matrix([[4, -3],[8,3]])
A3
```

```python
b3=sp.Matrix([7,3])
b3
```

```python
x_mal=A3.T@b3
x_mal
```

Tampoco sirve usar las proyecciones

```python
x0= b3.dot(A3[:,0])/A3[:,0].dot(A3[:,0])
x0
```

```python

x1= b3.dot(A3[:,1])/A3[:,1].dot(A3[:,1])
x1
```

Solo funciona resolver el sistema de ecuaciones

```python
M=A3[:,:].row_join(b3)
M
```

```python
M[1,:] -= M[0,:]*2
M
```

```python
M[0,:] += M[1,:]/3
M
```
